## Step 1: Import Required Libraries

We need two main libraries:
- `openai` - to communicate with GPT
- `dotenv` - to load our API key from the .env file


````python
# Import the libraries we need
from openai import OpenAI
from dotenv import load_dotenv
import os
````

## Step 2: Load Your API Key

Make sure you have created a `.env` file in the same folder as this notebook with:
```
OPENAI_API_KEY=sk-your-key-here
```


````python
# Load environment variables from .env file
load_dotenv()

# Get the API key from environment
api_key = os.getenv("OPENAI_API_KEY")

# Check if the key was loaded successfully
if api_key and api_key.startswith("sk-"):
    print("API key found and looks good!")
else:
    print("Error: API key not found or invalid. Please check your .env file.")
````

## Step 3: Your First API Call

Now let's make our first call to GPT! We'll send a simple greeting and see what comes back.

The format OpenAI expects is a list of dictionaries with `role` and `content` keys.


````python
# Create an OpenAI client instance
openai = OpenAI()

# Our first message to GPT
message = "Hello, GPT! This is my first ever message to you. Hi!"

# Format it as OpenAI expects - a list with a dictionary
messages = [
    {"role": "user", "content": message}
]

# Let's see what this looks like
print("Our messages list:")
print(messages)
````

````python
# Make the API call to GPT
# This is the Chat Completions API - you'll use this a lot!
response = openai.chat.completions.create(
    model="gpt-4o-mini",  # Using a fast, affordable model
    messages=messages
)

# Extract the response text
# The response comes back in: response.choices[0].message.content
reply = response.choices[0].message.content

print("GPT's response:")
print(reply)
````

## Step 4: Understanding System vs User Prompts

There are two types of messages you can send:
- **System prompt**: Sets up the AI's behavior, tone, and task
- **User prompt**: The actual question or input

Let's see how changing the system prompt changes the response!


````python
# Example 1: Helpful assistant
messages_helpful = [
    {"role": "system", "content": "You are a helpful assistant."},
    {"role": "user", "content": "What is 2 + 2?"}
]

response = openai.chat.completions.create(
    model="gpt-4o-mini",
    messages=messages_helpful
)

print("Helpful assistant says:")
print(response.choices[0].message.content)
````

````python
# Example 2: Snarky assistant - same question, different personality!
messages_snarky = [
    {"role": "system", "content": "You are a snarky assistant who answers correctly but with attitude."},
    {"role": "user", "content": "What is 2 + 2?"}
]

response = openai.chat.completions.create(
    model="gpt-4o-mini",
    messages=messages_snarky
)

print("Snarky assistant says:")
print(response.choices[0].message.content)
````

## Step 5: Putting It Together - A Reusable Function

Let's create a simple function that makes it easy to ask GPT questions with different personalities.


````python
def ask_gpt(user_question, system_prompt="You are a helpful assistant.", model="gpt-4o-mini"):
    """
    A simple function to ask GPT a question.
    
    Args:
        user_question: The question you want to ask
        system_prompt: How the AI should behave (optional)
        model: Which GPT model to use (optional)
    
    Returns:
        The AI's response as a string
    """
    # Build the messages list
    messages = [
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": user_question}
    ]
    
    # Make the API call
    response = openai.chat.completions.create(
        model=model,
        messages=messages
    )
    
    # Return just the text response
    return response.choices[0].message.content
````

````python
# Try it out! Simple question with default settings
answer = ask_gpt("What is the capital of France?")
print(answer)
````

````python
# Same question, but ask a pirate!
pirate_prompt = "You are a pirate. Answer questions while speaking like a pirate."
answer = ask_gpt("What is the capital of France?", system_prompt=pirate_prompt)
print(answer)
````
